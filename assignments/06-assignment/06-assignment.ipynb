1.  Sometimes experimental researchers exclude subjects from their
    analysis because the subjects (1) appear to understand what
    hypothesis the experiment is testing, (2) seem not to be taking the
    experiment seriously, or (3) fail to follow directions. Discuss
    whether each of these three practices is likely to introduce bias
    when the researcher compares average outcomes among non-excluded
    subjects. (From Gerber & Green, 2012, p. 245)

Excluding subjects from an experiment based on their behavior or perceived understanding can introduce bias, depending on the exclusion criteria. If participants discern the hypothesis and change their behavior accordingly, their inclusion could introduce bias. However, excluding them could also lead to bias if those who understand the hypothesis differ systematically from those who do not. For example, more educated or attentive participants may be more likely to infer the hypothesis, meaning the remaining sample may be less representative of the general population. Similarly, removing participants who do not take the experiment seriously can be beneficial for internal validity, ensuring that the observed effects are due to the intervention rather than random responses. However, if the definition of "not serious" disproportionately removes certain types of participants, such as those with lower motivation or specific personality traits, the remaining sample may not reflect the broader population. This introduces selection bias, potentially overestimating the intervention’s effectiveness in real-world settings.

Participants who fail to follow directions may introduce noise into the data rather than systematic bias. However, excluding them can introduce bias if noncompliance correlates with characteristics relevant to the treatment effects. For example, if individuals with lower cognitive ability or attention spans are more likely to be removed, the remaining sample might not generalize well to populations that include such individuals. While exclusions can sometimes improve the quality of an experiment, they risk introducing selection bias if the remaining sample is systematically different from the intended population. A better approach might involve robustness checks, intention-to-treat analyses, or sensitivity analyses to assess how exclusions influence results.

2.  Explain why relying solely on Inverse Probability Weighting (IPW) to
    adjust for attrition can be problematic. What assumptions are
    required for IPW to produce unbiased estimates? What are the
    consequences if these assumptions are violated? Is it better to
    present the estimate with and without IPW? Justify your answers with
    further readings Gerber & Green (2012, p. 244) has some suggested
    readings that you can use.

Relying solely on Inverse Probability Weighting (IPW) to adjust for attrition can be problematic because it assumes that missingness is entirely explained by observed covariates, meaning that, conditional on these covariates, the probability of attrition is unrelated to unobserved outcomes. This assumption, known as *Missing at Random (MAR)*, is strong and often unrealistic in real-world experiments where unobserved factors, such as motivation or health status, may influence both attrition and outcomes. If this assumption is violated, IPW can still produce biased estimates, particularly if the weights amplify noise by giving excessive influence to a small number of heavily weighted observations. Additionally, if the estimated probabilities of remaining in the study are very small for some individuals, the resulting weights may be highly variable, leading to unstable or inefficient estimates. To account for these potential issues, it is generally advisable to present estimates both with and without IPW to assess the sensitivity of the results. Researchers may also use other methods, such as bounding techniques (e.g., Lee bounds) or sensitivity analyses, to gauge the robustness of findings under different assumptions about the missing data mechanism. Gerber & Green (2012) suggest further readings such as Manski (1995) and Little & Rubin (2002), which discuss the limitations of IPW and alternative approaches to handling attrition bias.

4.  How is the validity of IPW estimates dependent on the assumption of
    MIPO\|X? Discuss what happens if this assumption is violated.

The validity of Inverse Probability Weighting (IPW) estimates relies on the assumption of *Missing at Random given X* (MIPO|X), which means that, conditional on observed covariates, the probability of missingness is independent of the unobserved potential outcomes. If this assumption holds, IPW can correct for attrition by reweighting the observed sample to resemble the full population, ensuring unbiased estimates. However, if this assumption is violated—meaning that missingness is correlated with unobserved factors affecting the outcome—IPW will fail to fully correct for selection bias. For example, if participants with worse health outcomes are more likely to drop out regardless of their observable characteristics, then weighting the remaining participants cannot account for this unmeasured difference, leading to biased treatment effect estimates. In such cases, sensitivity analyses, bounding methods, or alternative modeling approaches, such as selection models, may be needed to assess the extent of bias and explore how results change under different assumptions about the missing data mechanism.

2.  Estimate Manski bounds for the outcomes. Interpret the results.

```python
# Step 1: Identify treatment and control groups
treatment_group = df[df["treatment"] == "treatment"]
control_group = df[df["treatment"] == "control"]

# Step 2: Compute observed means (ignoring missing values)
observed_treatment_mean = treatment_group["outcome2"].mean(skipna=True)
observed_control_mean = control_group["outcome2"].mean(skipna=True)

# Step 3: Define worst and best possible values for outcome2
worst_outcome = df["outcome2"].min(skipna=True)  # Worst case scenario
best_outcome = df["outcome2"].max(skipna=True)   # Best case scenario

# Step 4: Compute Manski Bounds

# Lower Bound: Assign worst outcome to missing values
lower_bound_treatment = (treatment_group["outcome2"].sum(skipna=True) + worst_outcome * treatment_group["outcome2"].isna().sum()) / len(treatment_group)
lower_bound_control = (control_group["outcome2"].sum(skipna=True) + worst_outcome * control_group["outcome2"].isna().sum()) / len(control_group)
lower_bound_effect = lower_bound_treatment - lower_bound_control

# Upper Bound: Assign best outcome to missing values
upper_bound_treatment = (treatment_group["outcome2"].sum(skipna=True) + best_outcome * treatment_group["outcome2"].isna().sum()) / len(treatment_group)
upper_bound_control = (control_group["outcome2"].sum(skipna=True) + best_outcome * control_group["outcome2"].isna().sum()) / len(control_group)
upper_bound_effect = upper_bound_treatment - upper_bound_control

# Step 5: Display Results
manski_bounds_df = pd.DataFrame({
    "Group": ["Treatment", "Control"],
    "Observed Mean": [observed_treatment_mean, observed_control_mean],
    "Lower Bound (Worst Case)": [lower_bound_treatment, lower_bound_control],
    "Upper Bound (Best Case)": [upper_bound_treatment, upper_bound_control]
})

effect_bounds_df = pd.DataFrame({
    "Effect Type": ["Lower Bound", "Upper Bound"],
    "Treatment Effect Estimate": [lower_bound_effect, upper_bound_effect]
})

# Print results
print("Manski Bounds Estimates:")
print(manski_bounds_df)

print("\nManski Bounds for Treatment Effect:")
print(effect_bounds_df)
```

3.  Finally, estimate Lee bounds and contrast the result with the
    previous two. Which of the two methods would you use? Justify your
    answer and elaborate on the possible violation of the monotonicity
    assumption.

```python
def compute_lee_bounds(df, treatment_col, outcome_col, attrition_col):
    """
    Computes Lee Bounds for estimating treatment effect while accounting for attrition.
    
    Parameters:
    df (DataFrame): Dataset with treatment, outcome, and attrition variables.
    treatment_col (str): Column indicating treatment assignment.
    outcome_col (str): Outcome variable column.
    attrition_col (str): Attrition indicator column.

    Returns:
    lee_bounds_df (DataFrame): DataFrame with lower and upper Lee bound estimates.
    """
    # Step 1: Identify treatment and control groups
    treatment_group = df[df[treatment_col] == "treatment"]
    control_group = df[df[treatment_col] == "control"]

    # Step 2: Compute attrition rates
    attrition_rate_treatment = treatment_group[attrition_col].isna().mean()
    attrition_rate_control = control_group[attrition_col].isna().mean()

    # Step 3: Compute trimming proportion (use the group with the **lower attrition rate**)
    trim_prop = min(attrition_rate_treatment, attrition_rate_control)

    # Step 4: Trim the group with **higher retention** by removing observations with highest/lowest outcomes
    def trim_group(group, outcome_col, trim_prop):
        group_sorted = group.sort_values(by=outcome_col, ascending=True)
        n_trim = int(np.floor(len(group_sorted) * trim_prop))
        return group_sorted.iloc[n_trim:-n_trim] if n_trim > 0 else group_sorted
    
    if attrition_rate_treatment > attrition_rate_control:
        trimmed_treatment = trim_group(treatment_group, outcome_col, trim_prop)
        trimmed_control = control_group
    else:
        trimmed_treatment = treatment_group
        trimmed_control = trim_group(control_group, outcome_col, trim_prop)

    # Step 5: Compute Lee Bounds
    lower_bound_effect = trimmed_treatment[outcome_col].mean() - trimmed_control[outcome_col].mean()
    upper_bound_effect = treatment_group[outcome_col].mean(skipna=True) - control_group[outcome_col].mean(skipna=True)

    # Step 6: Store results in DataFrame
    lee_bounds_df = pd.DataFrame({
        "Bound Type": ["Lower Lee Bound", "Upper Lee Bound"],
        "Treatment Effect Estimate": [lower_bound_effect, upper_bound_effect]
    })
    
    return lee_bounds_df

# Apply Lee Bounds computation
lee_bounds_df = compute_lee_bounds(df, treatment_col="treatment", outcome_col="outcome2", attrition_col="outcome2")

# Print Results
print("Lee Bounds Estimates:")
print(lee_bounds_df)
```

Manski bounds provide the most conservative estimates by considering worst-case and best-case missing data scenarios, resulting in wider treatment effect bounds (0.83 to 1.5). In contrast, Lee bounds adjust for selection bias due to attrition by trimming excess observations, leading to narrower bounds that better reflect realistic treatment effects. The Lee bounds likely provide a more precise estimate of the treatment effect compared to the broad range given by Manski bounds. Between the two methods, I would prefer Lee bounds because they correct for systematic attrition rather than assuming extreme missing values. However, Lee bounds rely on the monotonicity assumption, which assumes that attrition is only affecting one group in a predictable way (e.g., treatment group more likely to remain). If monotonicity is violated—for example, if highly engaged individuals from both groups drop out—Lee bounds may underestimate the treatment effect bias. In such cases, Manski bounds are more appropriate since they do not depend on this assumption. Ultimately, the choice between the two depends on whether attrition is likely to be systematic (favoring Lee bounds) or unpredictable (favoring Manski bounds).
